```python
from myproject.utilities.iapi_TNG import get

baseUrl = 'http://www.tng-project.org/api/'
# headers = {"api-key":"env{apikey}"}
r=get(baseUrl)
names = [sim['name'] for sim in r['simulations']]
i = names.index(sim)
sim = get( r['simulations'][i]['url'] )
snaps = get( sim['snapshots'] )
```

```python
from myproject.utilities.joinTime import JoinTime
import os

rewrite=False

join_times_path = galaxyGroupData.scratchPlotDirc + f'/join_times_and_parameter_changes_{galaxyGroupData.sim}_{galaxyGroupData.snapshot_dic[galaxyGroupData.snapshot][1]}.txt'
satellites_without_tree_path = galaxyGroupData.scratchPlotDirc + f'/satellites_without_merger_tree_{galaxyGroupData.sim}_{galaxyGroupData.snapshot_dic[galaxyGroupData.snapshot][1]}.txt'

join_times_header = (
    "GalaxyGroupID\tSubhaloID\tNumMembers\tClusterMass\tJoiningRedshift\tSeparationZ0\tSeparationNorm\tDeltaMass\tDeltaTAll\tDeltaTLim\tDeltaMassTotal\tDeltaMassDM\tDeltaMassStars\tDeltaVsq\tJoinSnap\tClosestSnap\tClosestNorm\tClosestZ\tJoinProgSnap\tDeltaL\tSubhaloMassAtJoin\tHostProgID\tLJoin\tL0\tFirst1R200Redshift\tFirst2R200Redshift\tFirst3R200Redshift\n"
)

totalSatellites = sum(gg.getNumSubhalos() for gg in galaxyGroupData.filtered_and_corrected_list_galaxy_groups.getAllGalaxyGroups())
print(f"Total number of satellites to process (including centrals in counts): {totalSatellites}")

processedJoinTimes = []
satelliteToJoinTimeMap: dict[int, JoinTime] = {}

satellitesWihtoutMergerTree: set[int] = set()
#load satellitesWihtoutMergerTree from file if exists
# if rewrite==True:
if os.path.exists(satellites_without_tree_path):
    with open(satellites_without_tree_path, 'r') as f:
        for line in f:
            s = line.strip()
            if not s or s.lower().startswith('satellitesubhaloid'):
                continue
            try:
                satellitesWihtoutMergerTree.add(int(s.split('\t')[0]))
            except Exception:
                continue
    print(f"Loaded {len(satellitesWihtoutMergerTree)} satellites without merger tree from file.")
else:
    print("No existing file for satellites without merger tree found, starting with an empty list.")
    #prep the file with header
    with open(satellites_without_tree_path, 'w') as f:
        f.write("SatelliteSubhaloID\n")

# If not rewriting, resume from existing join-times file when available.
completed_satellite_ids: set[int] = set()
join_file_mode = 'w'
should_write_header = True
if not rewrite and os.path.exists(join_times_path):
    join_file_mode = 'a'
    should_write_header = os.path.getsize(join_times_path) == 0
    try:
        with open(join_times_path, 'r') as f:
            for line in f:
                s = line.strip()
                if not s or s.lower().startswith('galaxygroupid'):
                    continue
                parts = s.split('\t')
                if len(parts) < 2:
                    continue
                try:
                    completed_satellite_ids.add(int(parts[1]))
                except Exception:
                    continue
        print(
            f"Resuming from existing {os.path.basename(join_times_path)}: "
            f"{len(completed_satellite_ids)} satellites already completed."
        )
    except Exception as e:
        print(f"WARNING: Failed to read existing join-times file for resume ({e}); will append anyway.")


tasks: list[tuple] = []
for gg in galaxyGroupData.filtered_and_corrected_list_galaxy_groups.getAllGalaxyGroups():
    gg_id = gg.getGroupID()
    num_members = gg.getNumSubhalos()
    cluster_mass = gg.getMCrit200()
    central_subhalo = gg.getCentralSubhalo()
    if central_subhalo is None:
        continue

    host_id = central_subhalo.getIdx()
    for subhalo in gg.getSatelliteSubhalos():
        sat_id = subhalo.getIdx()
        if sat_id in satellitesWihtoutMergerTree:
            continue
        if sat_id in completed_satellite_ids:
            continue
        tasks.append(
            (
                galaxyGroupData.sim,
                galaxyGroupData.snapshot,
                host_id,
                sat_id,
                galaxyGroupData.sim_boxsize_kpc,
                galaxyGroupData.sim_boxsize_kpc/2,
                galaxyGroupData.scratchDataDirc + '/mergerTree',
                gg_id,
                num_members,
                cluster_mass,
            )
        )

```

```python

sim, snapshot, host_id, sat_id, L, halfbox, fname, gg_id, num_members, cluster_mass = tasks[0]
joinTime = JoinTime(sim, snapshot)
join_time_info = joinTime.computeJoinTimes(hostID=host_id, ID=sat_id, L=L, halfbox=halfbox, fname=fname)
print(join_time_info)
```
